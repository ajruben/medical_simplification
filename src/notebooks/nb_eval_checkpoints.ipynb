# Model Analysis for Medical Text Simplification

This notebook evaluates various fine-tuned models on the medical text simplification task.

**Goals:**
1. Load existing results from `analysis_results.csv`.
2. Identify all specified model paths.
3. For each model:
    a. If already evaluated (present in CSV), calculate only missing metrics (BLEU, Jaccard Similarity).
    b. If new, evaluate fully (BLEU, ROUGE, SARI, Jaccard Similarity).
4. Merge new/updated results with existing ones.
5. Save the combined results back to `analysis_results.csv`.

## 2. Load Data

```python
print("Loading datasets...")
try:
    data_files = {}
    if os.path.exists(TRAIN_FILE):
        data_files["train"] = TRAIN_FILE
    if os.path.exists(VALIDATION_FILE):
        data_files["validation"] = VALIDATION_FILE

    if not data_files:
        raise FileNotFoundError("No train or validation data files found.")

    raw_datasets = load_dataset("csv", data_files=data_files)

    # Rename columns if necessary
    renamed_datasets = DatasetDict()
    for split in raw_datasets:
        current_cols = raw_datasets[split].column_names
        temp_ds = raw_datasets[split]
        if "Expert" in current_cols and "Simple" in current_cols:
            temp_ds = temp_ds.rename_column("Expert", "input_text")
            temp_ds = temp_ds.rename_column("Simple", "target_text")
            print(f"Renamed columns in '{split}' split.")
        elif "input_text" not in current_cols or "target_text" not in current_cols:
             raise ValueError(f"Expected columns ('Expert', 'Simple' or 'input_text', 'target_text') not found in split '{split}'. Found: {current_cols}")
        renamed_datasets[split] = temp_ds

    print("Dataset loaded and columns checked/renamed.")
    print(renamed_datasets)

    # Select subsets for evaluation if configured
    if "validation" in renamed_datasets and NUM_SAMPLES_VALIDATION is not None:
        val_indices = range(min(NUM_SAMPLES_VALIDATION, len(renamed_datasets["validation"])))
        eval_validation_dataset = renamed_datasets["validation"].select(val_indices)
        print(f"Using subset of {len(eval_validation_dataset)} samples for validation evaluation.")
    elif "validation" in renamed_datasets:
        eval_validation_dataset = renamed_datasets["validation"]
        print(f"Using full validation set ({len(eval_validation_dataset)} samples) for evaluation.")
    else:
        eval_validation_dataset = None
        print("Warning: No validation dataset found or loaded.")

    if "train" in renamed_datasets and NUM_SAMPLES_TRAINING is not None:
        train_indices = range(min(NUM_SAMPLES_TRAINING, len(renamed_datasets["train"])))
        eval_train_dataset_for_similarity = renamed_datasets["train"].select(train_indices)
        print(f"Using subset of {len(eval_train_dataset_for_similarity)} samples from training set for similarity check.")
    elif "train" in renamed_datasets:
        eval_train_dataset_for_similarity = renamed_datasets["train"]
        print(f"Using full training set ({len(eval_train_dataset_for_similarity)} samples) for similarity check.")
    else:
        eval_train_dataset_for_similarity = None
        print("Warning: No training dataset found or loaded for similarity check.")

except Exception as e:
    print(f"Error loading or processing dataset: {e}")
    # Optionally raise e or handle differently
    eval_validation_dataset = None
    eval_train_dataset_for_similarity = None
```

## 3. Metrics Setup

```python
print("Loading evaluation metrics...")
try:
    sari_metric = evaluate.load("sari")
    bleu_metric = evaluate.load("bleu")
    rouge_metric = evaluate.load("rouge")
    print("Metrics loaded successfully.")
except Exception as e:
    print(f"Error loading metrics: {e}")
    # Handle error appropriately

def calculate_jaccard(text1, text2):
    """Calculates Jaccard similarity between the word sets of two texts."""
    if not text1 or not text2:
        return 0.0
    set1 = set(nltk.word_tokenize(text1.lower()))
    set2 = set(nltk.word_tokenize(text2.lower()))
    intersection = len(set1.intersection(set2))
    union = len(set1.union(set2))
    return intersection / union if union > 0 else 0.0

def preprocess_text_for_metric(texts):
    """Basic preprocessing for BLEU/ROUGE/SARI: tokenization and joining sentences."""
    if isinstance(texts, str):
        return "\n".join(nltk.sent_tokenize(texts.strip()))
    elif isinstance(texts, list):
        # Handle list of lists for references
        if texts and isinstance(texts[0], list):
             return [["\n".join(nltk.sent_tokenize(ref.strip())) for ref in ref_list] for ref_list in texts]
        else:
             return ["\n".join(nltk.sent_tokenize(text.strip())) for text in texts]
    return texts # Return as is if format is unexpected
```

## 4. Evaluation Loop

```python
new_results = [] # Store results from this run

if eval_validation_dataset is None and eval_train_dataset_for_similarity is None:
    print("Cannot proceed with evaluation as no datasets were loaded successfully.")
else:
    print(f"Starting evaluation loop for {len(model_dirs_to_evaluate)} models...")

    for model_dir_raw in tqdm(model_dirs_to_evaluate, desc="Evaluating Models"):
        model_dir = os.path.normpath(model_dir_raw)
        model_name = os.path.basename(model_dir) # Use normalized path for name
        print(f"\n--- Evaluating: {model_name} ({model_dir}) ---")

        # Check if model exists in existing results
        existing_row = None
        needs_bleu = True
        needs_rouge = True
        needs_sari = True
        needs_jaccard = True

        if existing_results_df is not None and 'model_path_normalized' in existing_results_df.columns:
            match = existing_results_df[existing_results_df['model_path_normalized'] == model_dir]
            if not match.empty:
                existing_row = match.iloc[0]
                print(f"Found existing results for {model_name}. Checking for missing metrics.")
                # Check which metrics are missing (NaN or not present)
                needs_bleu = pd.isna(existing_row.get('bleu', np.nan))
                needs_rouge = pd.isna(existing_row.get('rougeL', np.nan)) # Check one ROUGE score
                needs_sari = pd.isna(existing_row.get('sari', np.nan))
                needs_jaccard = pd.isna(existing_row.get('avg_jaccard_train_output', np.nan))
                print(f"Needs BLEU: {needs_bleu}, Needs ROUGE: {needs_rouge}, Needs SARI: {needs_sari}, Needs Jaccard: {needs_jaccard}")
            else:
                 print(f"No existing results found for {model_name}. Evaluating fully.")
        else:
             print(f"No existing results loaded or path column missing. Evaluating {model_name} fully.")

        # Skip if all required metrics are already present
        if not needs_bleu and not needs_rouge and not needs_sari and not needs_jaccard:
            print(f"Skipping {model_name} as all required metrics are present in existing results.")
            # Optionally copy existing results to new_results if you want them in the final merged df anyway
            # if existing_row is not None:
            #    new_results.append(existing_row.to_dict())
            continue

        # Initialize results for this model for this run
        current_run_results = {"model_name": model_name, "model_path": model_dir}
        model = None
        tokenizer = None

        try:
            # --- Load Model and Tokenizer ---
            print("Loading tokenizer...")
            tokenizer = AutoTokenizer.from_pretrained(model_dir)
            print("Loading model...")
            # Add specific loading logic if needed (e.g., for Mistral CausalLM)
            # if 'mistral' in model_dir.lower() and 'checkpoint' in model_dir.lower():
            #     print("Attempting to load Mistral checkpoint as CausalLM...")
            #     model = AutoModelForCausalLM.from_pretrained(model_dir, torch_dtype=torch.float16)
            # else:
            model = AutoModelForSeq2SeqLM.from_pretrained(model_dir)

            model.to(DEVICE)
            model.eval()
            print("Model and tokenizer loaded.")

            # --- Generate Predictions (only if any metric needs calculation) ---
            val_predictions_raw = None
            train_predictions_raw = None

            if (needs_bleu or needs_rouge or needs_sari) and eval_validation_dataset is not None:
                print("Generating predictions on validation set...")
                val_inputs = [SIMPLIFICATION_PREFIX + text for text in eval_validation_dataset['input_text']]
                val_predictions_raw = []
                with torch.no_grad():
                    for i in tqdm(range(0, len(val_inputs), BATCH_SIZE), desc=f"Validation Gen ({model_name})"):
                        batch_inputs = val_inputs[i:i+BATCH_SIZE]
                        inputs_tokenized = tokenizer(batch_inputs, return_tensors="pt", padding=True, truncation=True, max_length=MAX_LENGTH).to(DEVICE)
                        outputs_tokenized = model.generate(
                            **inputs_tokenized,
                            max_length=MAX_LENGTH,
                            num_beams=4, 
                            early_stopping=True
                        )
                        batch_preds = tokenizer.batch_decode(outputs_tokenized, skip_special_tokens=True)
                        val_predictions_raw.extend(batch_preds)

            if needs_jaccard and eval_train_dataset_for_similarity is not None:
                print("Generating predictions on training set for similarity...")
                train_inputs_for_gen = [SIMPLIFICATION_PREFIX + text for text in eval_train_dataset_for_similarity['input_text']]
                train_predictions_raw = []
                with torch.no_grad():
                    for i in tqdm(range(0, len(train_inputs_for_gen), BATCH_SIZE), desc=f"Training Sim Gen ({model_name})"):
                        batch_inputs = train_inputs_for_gen[i:i+BATCH_SIZE]
                        inputs_tokenized = tokenizer(batch_inputs, return_tensors="pt", padding=True, truncation=True, max_length=MAX_LENGTH).to(DEVICE)
                        outputs_tokenized = model.generate(
                            **inputs_tokenized,
                            max_length=MAX_LENGTH,
                            num_beams=4,
                            early_stopping=True
                        )
                        batch_preds = tokenizer.batch_decode(outputs_tokenized, skip_special_tokens=True)
                        train_predictions_raw.extend(batch_preds)

            # --- Calculate Metrics ---
            if val_predictions_raw is not None:
                val_references_raw = eval_validation_dataset['target_text']
                val_sources_raw = eval_validation_dataset['input_text']
                val_predictions_pp = preprocess_text_for_metric(val_predictions_raw)
                val_references_pp = preprocess_text_for_metric([[r] for r in val_references_raw])
                val_sources_pp = preprocess_text_for_metric(val_sources_raw)

                if needs_bleu:
                    try:
                        bleu_results = bleu_metric.compute(predictions=val_predictions_pp, references=val_references_pp)
                        current_run_results['bleu'] = bleu_results.get('bleu', 0.0)
                        print(f"  BLEU: {current_run_results['bleu']:.4f}")
                    except Exception as e:
                        print(f"  Error calculating BLEU: {e}")
                        current_run_results['bleu'] = None
                elif existing_row is not None: current_run_results['bleu'] = existing_row.get('bleu') # Carry over existing

                if needs_rouge:
                    try:
                        rouge_results = rouge_metric.compute(predictions=val_predictions_pp, references=val_references_pp)
                        current_run_results['rouge1'] = rouge_results.get('rouge1', 0.0)
                        current_run_results['rouge2'] = rouge_results.get('rouge2', 0.0)
                        current_run_results['rougeL'] = rouge_results.get('rougeL', 0.0)
                        print(f"  ROUGE-L: {current_run_results['rougeL']:.4f}")
                    except Exception as e:
                        print(f"  Error calculating ROUGE: {e}")
                        current_run_results['rouge1'] = current_run_results['rouge2'] = current_run_results['rougeL'] = None
                elif existing_row is not None: # Carry over existing
                    current_run_results['rouge1'] = existing_row.get('rouge1', existing_row.get('metric_rouge1')) # Check old format too
                    current_run_results['rouge2'] = existing_row.get('rouge2', existing_row.get('metric_rouge2'))
                    current_run_results['rougeL'] = existing_row.get('rougeL', existing_row.get('metric_rougeL'))

                if needs_sari:
                    try:
                        sari_results = sari_metric.compute(sources=val_sources_pp, predictions=val_predictions_pp, references=val_references_pp)
                        current_run_results['sari'] = sari_results.get('sari', 0.0)
                        print(f"  SARI: {current_run_results['sari']:.4f}")
                    except Exception as e:
                        print(f"  Error calculating SARI: {e}")
                        current_run_results['sari'] = None
                elif existing_row is not None: current_run_results['sari'] = existing_row.get('sari', existing_row.get('metric_sari')) # Carry over existing

            elif existing_row is not None: # No validation predictions generated, carry over all existing validation metrics
                 print("Carrying over existing validation metrics (BLEU, ROUGE, SARI).")
                 current_run_results['bleu'] = existing_row.get('bleu')
                 current_run_results['rouge1'] = existing_row.get('rouge1', existing_row.get('metric_rouge1'))
                 current_run_results['rouge2'] = existing_row.get('rouge2', existing_row.get('metric_rouge2'))
                 current_run_results['rougeL'] = existing_row.get('rougeL', existing_row.get('metric_rougeL'))
                 current_run_results['sari'] = existing_row.get('sari', existing_row.get('metric_sari'))
            else:
                 print("Skipping validation metric calculation (no data or predictions).")

            # --- Training Set Input Similarity ---
            if train_predictions_raw is not None:
                if needs_jaccard:
                    train_original_targets = eval_train_dataset_for_similarity['target_text']
                    jaccard_scores = []
                    for pred, target in zip(train_predictions_raw, train_original_targets):
                        jaccard_scores.append(calculate_jaccard(pred, target))

                    if jaccard_scores:
                        current_run_results['avg_jaccard_train_output'] = sum(jaccard_scores) / len(jaccard_scores)
                        print(f"  Avg Jaccard (Pred vs Train Target): {current_run_results['avg_jaccard_train_output']:.4f}")
                    else:
                        current_run_results['avg_jaccard_train_output'] = None
                        print("  Could not calculate Jaccard similarity (no results).")
                elif existing_row is not None: # Carry over existing
                    current_run_results['avg_jaccard_train_output'] = existing_row.get('avg_jaccard_train_output')
            elif existing_row is not None and not needs_jaccard: # No train predictions generated, carry over existing
                 print("Carrying over existing Jaccard similarity.")
                 current_run_results['avg_jaccard_train_output'] = existing_row.get('avg_jaccard_train_output')
            else:
                print("Skipping training set similarity evaluation (no data or predictions).")

            new_results.append(current_run_results)

        except Exception as e:
            print(f"ERROR evaluating model {model_name}: {e}")
            error_results = {"model_name": model_name, "model_path": model_dir, "error": str(e)}
            new_results.append(error_results)

        finally:
            # Clean up memory
            del model
            del tokenizer
            if torch.cuda.is_available():
                torch.cuda.empty_cache()
            gc.collect()
            print(f"Finished evaluating {model_name}. Memory cleaned.")

    print("\nEvaluation loop finished.")
```

## 5. Combine and Save Results

```python
print("Processing and merging results...")

final_results_df = None

if new_results:
    new_results_df = pd.DataFrame(new_results)
    # Add normalized path to new results for merging
    new_results_df['model_path_normalized'] = new_results_df['model_path'].apply(os.path.normpath)
    print(f"Generated {len(new_results_df)} new/updated results in this run.")

    if existing_results_df is not None:
        print("Merging new results with existing results...")
        # Ensure consistent columns before merging
        # Combine columns from both dataframes
        all_cols = list(existing_results_df.columns.union(new_results_df.columns))
        existing_results_df = existing_results_df.reindex(columns=all_cols)
        new_results_df = new_results_df.reindex(columns=all_cols)

        # Concatenate, prioritizing new results for the same model path
        combined_df = pd.concat([existing_results_df, new_results_df], ignore_index=True)
        # Keep the last entry for each model path (which will be the new one if it exists)
        final_results_df = combined_df.drop_duplicates(subset=['model_path_normalized'], keep='last')
        print(f"Merged DataFrame has {len(final_results_df)} rows.")
    else:
        # No existing results, the new results are the final results
        final_results_df = new_results_df
        print("Using new results as final results (no existing file found).")

elif existing_results_df is not None:
    # No new results generated, just use the existing ones
    final_results_df = existing_results_df
    print("No new results generated in this run. Using existing results.")
else:
    print("No results were generated or loaded.")

# --- Save Final Results --- 
if final_results_df is not None:
    # Define desired column order (including potential old columns)
    cols_order = [
        'model_name', 'model_path', # Core identifiers
        'bleu', 'rougeL', 'rouge1', 'rouge2', 'sari', # Main metrics
        'avg_jaccard_train_output', # Custom metric
        'error', # Error tracking
        # Add columns from the old CSV format if they exist
        'model_name_old', 'Sweep ID', 'Origin Folder', 'Dataset', 'Model Size', 'Metric Type',
        'metric_epoch', 'metric_loss', 'metric_rouge1', 'metric_rouge2', 'metric_rougeL',
        'metric_rougeLsum', 'metric_sari', 'metric_step', 'config_architectures',
        'config_d_model', 'config_name_or_path', 'config_num_heads', 'config_num_layers',
        'train_eval_batch_size', 'train_learning_rate', 'train_lr_scheduler',
        'train_num_epochs', 'train_optimizer', 'train_train_batch_size', 'train_weight_decay',
        'model_path_normalized' # Internal use, drop before saving
    ]

    # Get existing columns and reorder
    existing_cols = final_results_df.columns.tolist()
    final_cols_ordered = [col for col in cols_order if col in existing_cols] + \
                         [col for col in existing_cols if col not in cols_order and col != 'model_path_normalized'] # Add others, drop helper
    
    final_results_df_to_save = final_results_df[final_cols_ordered]

    print("\nFinal Combined Results DataFrame (Preview):")
    # Display might not work directly here, use print
    print(final_results_df_to_save.head())

    try:
        final_results_df_to_save.to_csv(OUTPUT_CSV, index=False, encoding='utf-8')
        print(f"\nResults successfully saved to {OUTPUT_CSV}")
    except Exception as e:
        print(f"\nError saving results to CSV: {e}")
else:
    print("No final results DataFrame to save.")
```
